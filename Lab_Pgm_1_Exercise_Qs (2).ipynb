```python
import numpy as np
import matplotlib.pyplot as plt
import itertools
```

```python
np.random.seed(42)

# --- PART 1A: Classical Probability ---
coin_space = ['H', 'T']
two_coin = list(itertools.product(coin_space, coin_space))
print('Sample Space:', two_coin)
```

```python
two_heads = [s for s in two_coin if s.count('H') == 2]
P_2H = len(two_heads) / len(two_coin)
print(f'P(2 Heads) = {P_2H}')
```

## PART 1A — Dice Addition Rule
### Now the program demonstrates the addition rule of probability.

```python
# Dice - Addition Rule
dice = list(range(1, 7))
A = [x for x in dice if x % 2 == 0]   # Even
B = [x for x in dice if x > 3]          # Greater than 3
P_A = len(A)/6;  P_B = len(B)/6
P_AB = len(set(A)&set(B))/6
print(f'P(A∪B) = P(A)+P(B)-P(A∩B) = {P_A+P_B-P_AB:.4f}')
```

## PART 1B: Empirical Probability
### Now the program moves from theoretical probability to experimental probability.

```python
# --- PART 1B: Empirical Convergence ---
for n in [10, 100, 1000, 10000]:
    tosses = np.random.binomial(1, 0.5, n)
    print(f'n={n:>6}: P(H)={tosses.mean():.4f}')
```

## PART 1C: Bayes' Theorem
### This is the most important statistical section of the program.
### The example considers a medical diagnostic test.

```python
# --- PART 1C: Bayes' Theorem ---
P_Disease = 0.01    # Only 1% of the population has the disease.
P_Pos_D   = 0.99    # Sensitivity - If a person has the disease, the probability that the test is positive is 99%.
P_Pos_ND  = 0.05    # False Positive Rate - If a person does NOT have the disease, there is still a 5% probability that the test is positive.


# Total Probability
P_Pos = P_Pos_D*P_Disease + P_Pos_ND*(1-P_Disease)
# Bayes' Theorem
P_D_Pos = (P_Pos_D * P_Disease) / P_Pos
print(f'P(Disease | Positive) = {P_D_Pos:.4f} ({P_D_Pos*100:.2f}%)')
```

# PART 2: Visualization
## The final section demonstrates the Law of Large Numbers visually.

```python
# --- VISUALIZATION ---
# Empirical convergence plot
n_total = 10000
tosses  = np.random.binomial(1, 0.5, n_total)
emp_prob = [tosses[:i].mean() for i in range(1, n_total+1)]
plt.figure(figsize=(10,4))
plt.plot(emp_prob, color='steelblue', linewidth=1)
plt.axhline(0.5, color='red', linestyle='--', label='True P=0.5')
plt.xscale('log')
plt.title('Empirical Probability Convergence (Law of Large Numbers)')
plt.xlabel('Number of Trials (log scale)')
plt.ylabel('Empirical Probability')
plt.legend(); plt.tight_layout(); plt.show()
```

## Inference from the Graph
### The graph demonstrates the Law of Large Numbers through repeated simulation of a fair coin toss.
- The red dashed line at 0.5 represents the theoretical probability of obtaining Heads.
- For a small number of trials, the empirical probability fluctuates significantly, ranging approximately from 0.25 to 0.67.
- As the number of trials increases, these fluctuations gradually decrease.
- After a few hundred trials, the empirical probability becomes increasingly close to 0.5.
- At around 1,000–10,000 trials, the curve remains very close to the theoretical probability of 0.5.
- Therefore, the graph confirms that as the number of independent trials increases, the empirical probability converges toward the theoretical probability.

The graph clearly demonstrates the Law of Large Numbers. For a small number of trials, the empirical probability of obtaining Heads varies considerably due to random fluctuations. However, as the number of trials increases, the empirical probability gradually converges toward the theoretical probability of 0.5. Hence, a larger number of trials provides a more reliable estimate of the true probability.

```python
coin_space = ['H', 'T']
three_coin_sample_space = list(itertools.product(coin_space, coin_space,coin_space))
print('Sample Space:', three_coin_sample_space)

# a) Getting exactly 3 Heads
exactly_3_heads = [s for s in three_coin_sample_space if s.count('H') == 3]
P_3H = len(exactly_3_heads) / len(three_coin_sample_space)
print(f'P(Exactly 3 Heads) = {P_3H:.4f}')

# b) Getting exactly 2 Heads
exactly_2_heads = [s for s in three_coin_sample_space if s.count('H') == 2]
P_2H = len(exactly_2_heads) / len(three_coin_sample_space)
print(f'P(Exactly 2 Heads) = {P_2H:.4f}')

# c) Getting at least 1 Head
at_least_1_head = [s for s in three_coin_sample_space if s.count('H') >= 1]
P_at_least_1H = len(at_least_1_head) / len(three_coin_sample_space)
print(f'P(At least 1 Head) = {P_at_least_1H:.4f}')
```

```python
# 2. Dice Probability

die_sample_space = list(range(1, 7))
print('Sample Space (Single Die):', die_sample_space)

# a) Probability of getting an even number
even_numbers = [d for d in die_sample_space if d % 2 == 0]
P_even = len(even_numbers) / len(die_sample_space)
print(f'P(Even Number) = {P_even:.4f}')

# b) Probability of getting an odd number
odd_numbers = [d for d in die_sample_space if d % 2 != 0]
P_odd = len(odd_numbers) / len(die_sample_space)
print(f'P(Odd Number) = {P_odd:.4f}')

# c) Probability of getting a number greater than 4
greater_than_4 = [d for d in die_sample_space if d > 4]
P_gt_4 = len(greater_than_4) / len(die_sample_space)
print(f'P(Number > 4) = {P_gt_4:.4f}')

# d) Probability of getting a number less than or equal to 3
less_than_eq_3 = [d for d in die_sample_space if d <= 3]
P_lte_3 = len(less_than_eq_3) / len(die_sample_space)
print(f'P(Number <= 3) = {P_lte_3:.4f}')
```

```python
# 3. Two-Dice Experiment

two_dice_sample_space = list(itertools.product(range(1, 7), range(1, 7)))
print('Sample Space (Two Dice):', two_dice_sample_space)
print(f'Total possible outcomes: {len(two_dice_sample_space)}')

# a) The sum is 7
sum_is_7 = [roll for roll in two_dice_sample_space if sum(roll) == 7]
P_sum_7 = len(sum_is_7) / len(two_dice_sample_space)
print(f'P(Sum is 7) = {P_sum_7:.4f}')

# b) The sum is greater than 8
sum_gt_8 = [roll for roll in two_dice_sample_space if sum(roll) > 8]
P_sum_gt_8 = len(sum_gt_8) / len(two_dice_sample_space)
print(f'P(Sum > 8) = {P_sum_gt_8:.4f}')

# c) Both dice show the same number
same_number = [roll for roll in two_dice_sample_space if roll[0] == roll[1]]
P_same_number = len(same_number) / len(two_dice_sample_space)
print(f'P(Both dice show same number) = {P_same_number:.4f}')

# d) At least one die shows 6
at_least_one_6 = [roll for roll in two_dice_sample_space if 6 in roll]
P_at_least_one_6 = len(at_least_one_6) / len(two_dice_sample_space)
print(f'P(At least one die shows 6) = {P_at_least_one_6:.4f}')
```

```python
# 4. Addition Rule (Two Events)

die_sample_space = list(range(1, 7))

# Define Event A: number is even
A_event = [d for d in die_sample_space if d % 2 == 0]
P_A = len(A_event) / len(die_sample_space)

# Define Event B: number is less than 3
B_event = [d for d in die_sample_space if d < 3]
P_B = len(B_event) / len(die_sample_space)

# Calculate A intersect B
A_intersect_B = [d for d in A_event if d in B_event]
P_A_intersect_B = len(A_intersect_B) / len(die_sample_space)

# Calculate P(A union B) using the addition rule
P_A_union_B = P_A + P_B - P_A_intersect_B

print(f'Sample Space (Single Die): {die_sample_space}')
print(f'Event A (even number): {A_event}, P(A) = {P_A:.4f}')
print(f'Event B (number less than 3): {B_event}, P(B) = {P_B:.4f}')
print(f'Event A intersect B: {A_intersect_B}, P(A intersect B) = {P_A_intersect_B:.4f}')
print(f'P(A union B) = P(A) + P(B) - P(A intersect B) = {P_A_union_B:.4f}')
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Set seed for identical reproducibility
np.random.seed(42)

# 2. Simulate 10,000 fair coin tosses (1 = Heads, 0 = Tails)
n_tosses = 10000
tosses = np.random.randint(0, 2, size=n_tosses)

# 3. Compute sequential cumulative probabilities
cumulative_heads = np.cumsum(tosses)
toss_numbers = np.arange(1, n_tosses + 1)
cumulative_probabilities = cumulative_heads / toss_numbers

# 4. Generate the visualization
plt.figure(figsize=(10, 6))

# Plot the simulated progression
plt.plot(toss_numbers, cumulative_probabilities, label='Simulated Cumulative Probability', color='#1f77b4', lw=1.5)

# Plot the theoretical probability horizon line
plt.axhline(y=0.5, color='r', linestyle='--', label='Theoretical Probability P(H) = 0.5', lw=2)

# Configure logarithmic scaling and axis boundaries
plt.xscale('log')
plt.xlim(1, n_tosses)
plt.ylim(0, 1.0)

# Add titles, labels, and clean formatting
plt.title('Law of Large Numbers Simulation (10,000 Coin Tosses)', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Number of Tosses (Log Scale)', fontsize=12)
plt.ylabel('Cumulative Probability of Heads', fontsize=12)
plt.grid(True, which="both", linestyle=':', alpha=0.5)
plt.legend(loc='upper right', fontsize=11)

# Display the plot
plt.tight_layout()
plt.show()
```

```python
# 6. Conditional Probability with Dice

die_sample_space = list(range(1, 7))

# Define Event A: number is even
A_event = [d for d in die_sample_space if d % 2 == 0]

# Define Event B: number is greater than 3
B_event = [d for d in die_sample_space if d > 3]

# Calculate A intersect B
A_intersect_B = [d for d in A_event if d in B_event]

# P(B) - Probability of B (number greater than 3)
P_B = len(B_event) / len(die_sample_space)

# P(A intersect B) - Probability of A and B (even and greater than 3)
P_A_intersect_B = len(A_intersect_B) / len(die_sample_space)

# Calculate P(A|B) = P(A intersect B) / P(B)
if P_B > 0:
    P_A_given_B = P_A_intersect_B / P_B
else:
    P_A_given_B = 0 # Or handle error as appropriate

print(f'Sample Space (Single Die): {die_sample_space}')
print(f'Event A (number is even): {A_event}')
print(f'Event B (number is greater than 3): {B_event}')
print(f'Event A intersect B: {A_intersect_B}')
print(f'P(A intersect B) = {P_A_intersect_B:.4f}')
print(f'P(B) = {P_B:.4f}')
print(f'P(A|B) = {P_A_given_B:.4f}')

# Manual Verification:
# A = {2, 4, 6}
# B = {4, 5, 6}
# A intersect B = {4, 6}
# P(A intersect B) = 2/6 = 1/3
# P(B) = 3/6 = 1/2
# P(A|B) = (1/3) / (1/2) = 2/3 = 0.6667
print('\nManual Verification:')
print('A = {2, 4, 6}')
print('B = {4, 5, 6}')
print('A intersect B = {4, 6}')
print('P(A intersect B) = 2/6 = 1/3')
print('P(B) = 3/6 = 1/2')
print('P(A|B) = (1/3) / (1/2) = 2/3 = 0.6667')
```

```python
# 7. Bayes' Theorem - Medical Test Problem

# Given probabilities
P_Disease = 0.02              # P(Disease): Prevalence of the disease (2%)
Sensitivity = 0.95            # P(Positive | Disease): Sensitivity of the test (95%)
False_Positive_Rate = 0.04    # P(Positive | No Disease): False-positive rate (4%)

# Calculate P(No Disease)
P_No_Disease = 1 - P_Disease

# Calculate P(Positive | No Disease)
P_Positive_given_No_Disease = False_Positive_Rate

# Calculate P(Positive)
# P(Positive) = P(Positive | Disease) * P(Disease) + P(Positive | No Disease) * P(No Disease)
P_Positive = (Sensitivity * P_Disease) + (P_Positive_given_No_Disease * P_No_Disease)

# Calculate P(Disease | Positive) using Bayes' Theorem
# P(Disease | Positive) = [P(Positive | Disease) * P(Disease)] / P(Positive)
P_Disease_given_Positive = (Sensitivity * P_Disease) / P_Positive

print(f'P(Disease) = {P_Disease:.4f}')
print(f'P(Positive | Disease) = {Sensitivity:.4f}')
print(f'P(Positive | No Disease) = {False_Positive_Rate:.4f}')
print(f'Calculated P(Positive) = {P_Positive:.4f}')
print(f'P(Disease | Positive) = {P_Disease_given_Positive:.4f} ({P_Disease_given_Positive*100:.2f}%)')
```

```python
# 8. Email Classification

# Given probabilities
P_Spam = 0.20                 # P(Spam): 20% of emails are spam
P_Offer_given_Spam = 0.80     # P(Offer | Spam): Spam email contains 'offer' with 80% probability
P_Offer_given_NoSpam = 0.10   # P(Offer | No Spam): Non-spam email contains 'offer' with 10% probability

# Calculate P(No Spam)
P_NoSpam = 1 - P_Spam

# Calculate P(Offer)
# P(Offer) = P(Offer | Spam) * P(Spam) + P(Offer | No Spam) * P(No Spam)
P_Offer = (P_Offer_given_Spam * P_Spam) + (P_Offer_given_NoSpam * P_NoSpam)

# Calculate P(Spam | Offer) using Bayes' Theorem
# P(Spam | Offer) = [P(Offer | Spam) * P(Spam)] / P(Offer)
P_Spam_given_Offer = (P_Offer_given_Spam * P_Spam) / P_Offer

print(f'P(Spam) = {P_Spam:.4f}')
print(f'P(Offer | Spam) = {P_Offer_given_Spam:.4f}')
print(f'P(Offer | No Spam) = {P_Offer_given_NoSpam:.4f}')
print(f'Calculated P(Offer) = {P_Offer:.4f}')
print(f'P(Spam | Offer) = {P_Spam_given_Offer:.4f} ({P_Spam_given_Offer*100:.2f}%)')
```
